### TASK  3: Visualize the relationship between success rate of each orbit type


Next, we want to visually check if there are any relationship between success rate and orbit type.


Let's create a `bar chart` for the sucess rate of each orbit


Analyze the plotted bar chart to identify which orbits have the highest success rates.


```python
# HINT use groupby method on Orbit column and get the mean of Class column

df_grp = df.groupby('Orbit')['Class'].mean().reset_index()

# plt.figure(figsize = (15, 10))
sns.catplot(data=df_grp, x="Orbit", y="Class", kind="bar", aspect=2)
plt.title('Bar Chart of Orbit Type Vs Success Rate')
plt.xlabel("Orbit")
plt.ylabel("Success Rate")
plt.xticks(rotation=45)
plt.show()

# df_grp = df.groupby(['Orbit','Class']).size().unstack(fill_value=0).reset_index()

# df_grp.columns = df_grp.columns.astype(str)
# df_grp.rename(columns={
#     '0' : 'Fail',
#     '1' : 'Success'
# }, inplace=True)


# df_grp['Total'] = df_grp['Success'] + df_grp['Fail']
# df_grp['Success_Rate'] = df_grp['Success']/df_grp['Total']

# print(df_grp)


# # Create a figure with 1 row, 2 columns of subplots
# fig, axes = plt.subplots(1, 2, figsize=(15, 5))  # 1 row, 2 columns

# # ---- Left subplot: Counts of Fail/Success ----
# df_grp.set_index('Orbit')[['Fail', 'Success']].plot(
#     kind='bar',
#     ax=axes[0]  # send this plot to the first subplot
# )
# axes[0].set_title("Launch Outcome Counts by Orbit")
# axes[0].set_xlabel("Orbit Type")
# axes[0].set_ylabel("Count")
# axes[0].tick_params(axis='x', rotation=45)

# # ---- Right subplot: Success Rate ----
# df_grp.plot(
#     x='Orbit',
#     y='Success_Rate',
#     kind='bar',
#     ax=axes[1],  # send this plot to the second subplot
#     legend=False,
#     color='green'
# )
# axes[1].set_title("Success Rate by Orbit")
# axes[1].set_ylabel("Success Rate")
# axes[1].set_ylim(0, 1)
# axes[1].tick_params(axis='x', rotation=45)

# plt.tight_layout()  # adjust spacing
# plt.show()
```

```python
# High success rate of 1 for rare orbits like HEO, ES-L1 doesnt mean that they have better chances of successful landing, 
# the data sample could be too small so its helpful to view bar chart with counts on succeded and failed launches
# We are creating two subplots here 1st one shows a bar chart with failed/succeeded launch outcomes and
# 2nd one is the one with Success Rates. 2 plots can be viewwed side by side

df_grp = df.groupby(['Orbit','Class']).size().unstack(fill_value=0).reset_index()

df_grp.columns = df_grp.columns.astype(str)
df_grp.rename(columns={
    '0' : 'Fail',
    '1' : 'Success'
}, inplace=True)


df_grp['Total'] = df_grp['Success'] + df_grp['Fail']
df_grp['Success_Rate'] = df_grp['Success']/df_grp['Total']

print(df_grp)


# Create a figure with 1 row, 2 columns of subplots
fig, axes = plt.subplots(1, 2, figsize=(15, 5))  # 1 row, 2 columns

# ---- Left subplot: Counts of Fail/Success ----
df_grp.set_index('Orbit')[['Fail', 'Success']].plot(
    kind='bar',
    ax=axes[0]  # send this plot to the first subplot
)
axes[0].set_title("Launch Outcome Counts by Orbit")
axes[0].set_xlabel("Orbit Type")
axes[0].set_ylabel("Count")
axes[0].tick_params(axis='x', rotation=45)

# ---- Right subplot: Success Rate ----
df_grp.plot(
    x='Orbit',
    y='Success_Rate',
    kind='bar',
    ax=axes[1],  # send this plot to the second subplot
    legend=False,
    color='green'
)
axes[1].set_title("Success Rate by Orbit")
axes[1].set_ylabel("Success Rate")
axes[1].set_ylim(0, 1)
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()  # adjust spacing
plt.show()
```

### TASK  4: Visualize the relationship between FlightNumber and Orbit type


For each orbit, we want to see if there is any relationship between FlightNumber and Orbit type.


```python
# Plot a scatter point chart with x axis to be FlightNumber and y axis to be the Orbit, and hue to be the class value

ax = sns.catplot(data=df, x="FlightNumber", y="Orbit", hue='Class', aspect=2)
sns.move_legend(ax, "center right", frameon=True)
plt.title('Categorical Plot of Flight Number Vs Orbit Type')
plt.xlabel("FlightNumber",fontsize=15)
plt.ylabel("Orbit",fontsize=15)
plt.show()
```

### TASK  5: Visualize the relationship between Payload Mass and Orbit type


Similarly, we can plot the Payload Mass vs. Orbit scatter point charts to reveal the relationship between Payload Mass and Orbit type


```python
# Plot a scatter point chart with x axis to be Payload Mass and y axis to be the Orbit, and hue to be the class value
ax = sns.catplot(data=df, x="PayloadMass", y="Orbit", hue='Class', aspect=2)
sns.move_legend(ax, "center right", frameon=True)
plt.title('Categorical Plot of Payload Mass Vs Orbit Type')
plt.xlabel("Payload Mss(Kg)",fontsize=15)
plt.ylabel("Orbit",fontsize=15)
plt.show()
```

### TASK  6: Visualize the launch success yearly trend


You can plot a line chart with x axis to be <code>Year</code> and y axis to be average success rate, to get the average launch success trend.


The function will help you get the year from the date:


```python
# A function to Extract years from the date 
year=[]
def Extract_year():
    for i in df["Date"]:
        year.append(i.split("-")[0])
    return year
Extract_year()
df['Date'] = year
df.head()
    
```

```python
# Plot a line chart with x axis to be the extracted year and y axis to be the success rate

df_grouped = df.groupby('Date')['Class'].mean().reset_index()
plt.figure(figsize=(8, 6))
sns.lineplot(data=df_grouped, x='Date', y='Class')
plt.title('Line Chart of Year Vs Success Rate')
plt.xlabel('Date')
plt.ylabel('Average Sucess Rate')

plt.show()
```

you can observe that the sucess rate since 2013 kept increasing till 2020


## Features Engineering


By now, you should obtain some preliminary insights about how each important variable would affect the success rate, we will select the features that will be used in success prediction in the future module.


```python
features = df[['FlightNumber', 'PayloadMass', 'Orbit', 'LaunchSite', 'Flights', 'GridFins', 'Reused', 'Legs', 'LandingPad', 'Block', 'ReusedCount', 'Serial']]
features.head()
```

### TASK  7: Create dummy variables to categorical columns


Use the function <code>get_dummies</code> and <code>features</code> dataframe to apply OneHotEncoder to the column <code>Orbits</code>, <code>LaunchSite</code>, <code>LandingPad</code>, and <code>Serial</code>. Assign the value to the variable <code>features_one_hot</code>, display the results using the method head. Your result dataframe must include all features including the encoded ones.


```python
# HINT: Use get_dummies() function on the categorical columns

df_encoded = pd.get_dummies(features, columns=['Orbit', 'LaunchSite', 'LandingPad', 'Serial'], dtype=int)

df_encoded.columns
```

```python
# HINT: use astype function


df_encoded = df_encoded.astype('float64')

df_encoded.shape
```
